###  mean-filled columns
These 15 sensor columns have little skew and fewer than 1% of records outside their 1.5×IQR fences. Missing values are not replaced with zero.

```python
mean_columns = ['ambient_temperature', 'humidity', 'air_pressure',
                'motor_temperature', 'spindle_speed', 'motor_current', 'torque',
                'vibration', 'acoustic_level', 'bearing_temperature',
                'process_temperature', 'hydraulic_pressure',
                'coolant_temperature', 'material_feed_rate', 'feed_pressure']
mean_values = df[mean_columns].mean()
```

```python
df[mean_columns] = df[mean_columns].fillna(mean_values)
```

### Fill flow_rate with its column median
The median is calculated from the available flow_rate readings in this dataset.

```python
flow_median = df['flow_rate'].median()
df['flow_rate'] = df['flow_rate'].fillna(flow_median)
print('flow_rate fill value:', flow_median)
```

###  category gaps with the column mode


```python
category_columns = ['production_line', 'shift', 'product_type', 'operation_mode']
category_values = df[category_columns].mode().iloc[0]
df[category_columns] = df[category_columns].fillna(category_values)
```

### Show what was filled
This table lists the method, column fill value and number of missing cells that were replaced. No existing measurements are changed.

```python
fill_values = mean_values.copy()
fill_values['flow_rate'] = flow_median
fill_report = pd.DataFrame({'fill_value': fill_values, 'missing_before': missing_before[fill_values.index]})
fill_report['method'] = 'column mean'
fill_report.loc['flow_rate', 'method'] = 'column median'
fill_report
```

```python
fill_report.to_csv('imputation_report.csv', index_label='column')
print('Missing cells after filling:', df.isna().sum().sum())
```
